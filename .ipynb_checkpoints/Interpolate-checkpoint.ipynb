# Interpolation tool

This is a QC (Quality Control) driven data interpolation tool for UK-Flow-15 time-series flow data.  
This time series flow data contains ...

The code reads CSV files containing sensor readings every 15 minutes, finds data points flagged as poor quality, replaces them with interpolated values, and saves the cleaned data.

### Quality Control Flags
Each timestep is assigned a three-digit QC code as part of the quality control procedure. Each digit represents the result of an independent check:
 
- (X): Consistency check against other established UK hydrological products 
- (Y): Standard QC anomaly checks 
- (Z): High-flow event checks 

Within each check, the digit value indicates the severity of the issue — for example, a 0 might mean the data passed cleanly, while higher digits indicate progressively worse problems. We will expain this further later.

![QC Code](QC_Code.png)

### Filtering by QC Flag
To filter out bad data, you specify two things:  

- qc_column — which check to look at (0=X, 1=Y, 2=Z)
- qc_flags — which digit values in that position should be treated as bad data

For example, qc_column=1, qc_flags="38" would flag any timestep where the Y check returned a 3 or an 8:

"137" → Y digit is 3 → flagged ✗  
"060" → Y digit is 6 → not flagged ✓  
"238" → Y digit is 3 → flagged ✗  

Flagged values are then replaced by time-based interpolation from the surrounding good data.

### Data Format:
Each station has a CSV file with these columns:

- Datetime — timestamp (used as the index)  
- Value — the flow measurement at the station in m3s-1
- Resolution - the original temporal interval between each observation in minutes. Regular intervals are indicated numerically, for instance, 120 representing a two-hour interval. Special cases include irregular time intervals, which are coded as 0, and intervals with an anomalous one-second gap, coded as 1   
- Flag — a quality control code, e.g. 060 or 137

```python
# --- Standardises QC flag values to zero-padded 3-digit strings, so 60 becomes "060", 7 becomes "007", etc.---
def normalize_flag_to_3char(flag_series: pd.Series) -> pd.Series:
    """
    Normalize flag values to zero-padded 3-character strings:
      0   -> '000'
      60  -> '060'
      137 -> '137'
    Non-numeric or missing -> 0 -> '000'
    """
    nums = pd.to_numeric(flag_series, errors="coerce").fillna(0).astype(int)
    return nums.astype(str).str.zfill(3)


flag_str = normalize_flag_to_3char(df["flag"])
df["flag_3char"] = normalize_flag_to_3char(df["flag"])

# --- Build selection mask ---
selection_mask = build_qc_mask(flag_str, qc_column=qc_column, qc_flags=qc_flags)
```

```python
nums = pd.to_numeric(df["flag"], errors="coerce").fillna(0).astype(int)
nums = nums.astype(str).str.zfill(3)
```

###  Run Interpolation 

```python
# Flagged values are set to null, then filled in using time-based linear interpolation
# This estimates a value based on the surrounding timestamps, not just neighbouring row positions.
# Results are optionally rounded.

# --- Build interpolation log (old values) ---
interp_log = pd.DataFrame(columns=["station", "datetime", "old_value", "new_value"])
if selection_mask.any():
    selected_pts = (
        df.loc[selection_mask, ["value"]]
        .rename(columns={"value": "old_value"})
        .reset_index())
    
    # Prepare for interpolation
    df = df.sort_index()
    if not df.index.is_unique:
        # defensively collapse duplicates before time interpolation
        df = df.groupby(level=0).mean(numeric_only=True)

    # Null selected values and interpolate
    df.loc[selection_mask, "value"] = pd.NA
    df["value"] = df["value"].interpolate(method="time")

    # Round if requested
    if (round_decimals is not None) and (round_decimals >= 0):
        df["value"] = df["value"].round(round_decimals)

    # New values back into log
    selected_pts["new_value"] = selected_pts["datetime"].map(df["value"])
    selected_pts["station"] = station_id
    interp_log = selected_pts[["station", "datetime", "old_value", "new_value"]]

# --- Save cleaned data ---
# os.makedirs(out_dir_data, exist_ok=True)
# path_out = os.path.join(out_dir_data, f"{station_id}.csv")
# df.to_csv(path_out)

# # --- Append interpolation log if requested ---
# if out_log_csv and not interp_log.empty:
#     os.makedirs(os.path.dirname(out_log_csv) or ".", exist_ok=True)
#     if os.path.exists(out_log_csv) and os.path.getsize(out_log_csv) > 0:
#         interp_log.to_csv(out_log_csv, mode="a", header=False, index=False)
#     else:
#         interp_log.to_csv(out_log_csv, index=False)
```
