# Introduction: UK-FLOW15

**UK-FLOW15** is a quality-controlled, **15-minute** national-scale river flow dataset for the UK, comprising observations from 1,369 gauging stations across England, Scotland, Wales, and Northern Ireland.

Unlike some datasets that include only the cleanest records, UK-FLOW15 has been published in full — every available station and time step is included, regardless of data quality. This is a deliberate decision: data artefacts that are problematic for one type of analysis may be entirely irrelevant for another. For example, low-flow truncation errors do not influence flood frequency analysis, and removing such records would unnecessarily limit the dataset's usefulness.
    
This means that users of UK-Flow 15 must make decisions about which observations are suitable for their analysis, which values to exclude and which to correct. To support users in making informed decisions about data quality, each observation is accompanied by a **QC flag** that describes any issues identified during quality control. 
    
Data is available here: https://catalogue.ceh.ac.uk/datastore/eidchub/211710ac-f01b-4b52-807f-373babb1c368/  
Further information on the dataset is here: https://essd.copernicus.org/preprints/essd-2026-152/essd-2026-152.pdf

---


# Understanding the QC flags

Each observation in UK-FLOW15 is assigned a three-digit QC flag. Each digit corresponds to a separate quality control dimension, as illustrated below:

<!-- ![QC Code](InputFigures/QC_Code.png) -->

| Flag | Dimension | Description |
|---|---|---|
| **X** | Consistency check | Comparison against other established UK hydrological products (NRFA)|
| **Y** | Anomaly check | Standard QC checks for statistical anomalies |
| **Z** | High-flow check | Checks applied during high-flow events |

---    

For each of the three checks, the digit value indicates the nature of the issue. The tables below list the meaning of the range of possible values for each of the three checks.

```python
def find_truncated_flows(
    df,
    time_col="datetime",
    value_col="values",
    min_run=4,            # min consecutive identical values to count as "truncated"
    tol=0.0,              # absolute tolerance for "identical" (e.g. 0.001 for rounded data)
    high_quantile=0.95,   # runs at/above this quantile are labelled as high-flow (likely ceiling)
    ignore_zero=True,     # zero-flow runs are usually genuine dry spells, so skip them
    only_high=False,      # True -> only keep runs at high flows (peak clipping)
):
    df_local = df.reset_index().copy()
    d = (df_local[[time_col, value_col]].dropna().sort_values(time_col).reset_index(drop=True))
    t = d[time_col]
    v = d[value_col].to_numpy(float)

    # A run continues if value is unchanged AND there's no time gap
    step = t.diff().median()
    contiguous = (t.diff().iloc[1:] <= 1.5 * step).to_numpy()
    same = np.abs(np.diff(v)) <= tol
    new_run = np.ones(len(d), bool)
    new_run[1:] = ~(same & contiguous)
    d["run"] = np.cumsum(new_run)
    d["i"] = np.arange(len(d))

    runs = (d.groupby("run")
              .agg(start=(time_col, "first"), end=(time_col, "last"),
                   i0=("i", "first"), i1=("i", "last"),
                   n=(value_col, "size"), value=(value_col, "first"))
              .reset_index(drop=True))

    high_thr = np.quantile(v, high_quantile)
    runs = runs[runs["n"] >= min_run].copy()
    if ignore_zero:
        runs = runs[runs["value"] > 0]
    runs["at_high_flow"] = runs["value"] >= high_thr
    runs["is_global_max"] = np.isclose(runs["value"], v.max(), atol=max(tol, 1e-12))
    if only_high:
        runs = runs[runs["at_high_flow"]]
    runs["duration"] = runs["end"] - runs["start"]
    runs = runs.sort_values(["value", "n"], ascending=False).reset_index(drop=True)

    # Repeated hits on the same high value across separate runs = strong ceiling evidence
    hi = runs[runs["at_high_flow"]]
    if len(hi):
        counts = hi["value"].round(6).value_counts()
        repeated = counts[counts > 1]
        if len(repeated):
            print("Same high value hit in multiple separate runs (likely a ceiling):")
            # print(repeated.to_string(), "\n")

    plot_truncated_flows(runs, d, t, v, plot=True, max_panels= 10)            
            
    return runs, repeated

def plot_truncated_flows(runs, d, t, v, plot=True,
                        max_panels=6,         # number of zoomed panels
                        pad=24,               # samples of context either side in zoomed panels
                         high_quantile=0.95
                         ):
    
    high_thr = np.quantile(v, high_quantile)
    
    
    n_pan = min(max_panels, len(runs))
                         
    ncols = 2
    nrows = 1 + int(np.ceil(n_pan / ncols))
    fig = plt.figure(figsize=(14, 3 + 2.6 * (nrows - 1)))
    gs = fig.add_gridspec(nrows, ncols)

    # Overview
    ax = fig.add_subplot(gs[0, :])
    ax.plot(t, v, lw=0.6, color="0.3")
    mask = np.zeros(len(d), bool)
    for _, r in runs.iterrows():
        mask[r.i0:r.i1 + 1] = True
        ax.axvspan(r.start, r.end, color="red", alpha=0.25, lw=0)
    ax.scatter(t[mask], v[mask], s=6, color="red", zorder=3,
               label=f"truncated samples ({len(runs)} runs)")
    ax.axhline(high_thr, ls="--", color="orange", lw=0.8,
               label=f"{high_quantile:.0%} quantile")
    ax.set_title("Flow record with suspected truncated periods")
    ax.set_ylabel("value")
    ax.legend(loc="upper right")

    # Zoomed panels
    for k, (_, r) in enumerate(runs.head(n_pan).iterrows()):
        ax = fig.add_subplot(gs[1 + k // ncols, k % ncols])
        a, b = max(r.i0 - pad, 0), min(r.i1 + pad, len(d) - 1)
        ax.plot(t[a:b + 1], v[a:b + 1], "-o", ms=3, lw=0.8, color="0.3")
        ax.plot(t[r.i0:r.i1 + 1], v[r.i0:r.i1 + 1], "o", ms=5, color="red")
        ax.axhline(r.value, ls=":", color="red", lw=0.8)
        ax.set_title(f"{r.start:%Y-%m-%d %H:%M} | {r.n} steps | value={r.value:g}",
                     fontsize=9)
        ax.tick_params(axis="x", rotation=30, labelsize=8)

    fig.tight_layout()
    plt.show()



# Usage
runs, repeats = find_truncated_flows(df, time_col="datetime", value_col="value", min_run=4)
# runs.head()
```

Conclusion Diagnostics 
</h1>

<p style="color: #DCE7F5; margin: 0px 0px 12px 0px; font-size: 14px;">

- The prolonged truncations in March–April 2001 are unusual compared with the rest of the station record.  
- Agreement with NRFA peak flows, where established, provides confidence in the recorded magnitudes, but sub-daily variation may be poorly represented.  
- Use the truncation flags as a caution when analysing hydrograph shape during this period.
